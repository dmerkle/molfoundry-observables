# All examples

The [CLI guide](../tutorials/README.md) starts with the graph tutorial, then ABA and molecular ROP.

Start the notebooks with **[ABA](examples/02_aba.ipynb)**. Change `MOTIF` from `ABA` to `ABBA`, select the **MolFoundry** kernel, then **Restart Kernel and Run All**. Each notebook runs independently; code cells can be folded in JupyterLab.

On GitHub this is a preview of code and saved results. [Open an executable environment](https://github.com/dmerkle/molfoundry-observables#install-and-open).

| Query-dependent discovery | [ABA](examples/02_aba.ipynb), [query variation](examples/08_query_variation.ipynb) |
